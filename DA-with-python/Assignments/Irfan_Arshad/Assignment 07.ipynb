10. Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows

```python
employees = pd.DataFrame({
    'Name': ['Irfan','Arshad','Ahmed','Zain'],
    'Salary': [30000,50000,25000,70000],
    'DeptID': [1, 2, 1, 3]
})

departments = pd.DataFrame({
    'DeptID': [1, 2, 4],
    'Department': ['Data Analytics', 'Math', 'Finance']
})

inner = pd.merge(employees, departments, on='DeptID', how='inner')
print(inner)

left = pd.merge(employees, departments, on='DeptID', how='left')
print(left)

combined = pd.concat([employees, employees], ignore_index=True)

combined = combined.drop_duplicates()
print(combined)
```
