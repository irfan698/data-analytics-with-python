2. Create a 3x4 array of zeros, a 3x4 array of ones, and 4x4 identity matrix. Then create an array of 8 evenly spaced numbers between 0 to 1 using linespace.

```python

zeros = np.zeros((3, 4))
ones = np.ones((3, 4))
iden = np.eye(4)
num = np.linspace(0, 1, 8)

print("Zeros\n", zeros)
print("Ones\n", ones)
print("Identity matrix\n", iden)
print("evenly spaced numbers\n", num)
```

3. Create a 1D array of numbers from 1 to 24 and reshape it into 2D array of shape (4,6). Then reshape it into a 3D array of shape (2,3,4) and print shape of each.

```python

arr2 = np.arange(1, 25)
arr_2d = arr2.reshape(4, 6)
arr_3d = arr2.reshape(2, 3, 4)

print("1D array", arr2)
print("1D shape", arr2.shape)

print("\n2D array\n", arr_2d)
print("2D shape", arr_2d.shape)

print("\n3D array\n", arr_3d)
print("3D shape:", arr_3d.shape)
```

5. Create a 2D array of reshape(5,5) with values 1 to 25. extract (a) the second row, (b) the third column,(c) the 3x3 sub-matrix from the center.

```python
arr4 = np.arange(1,26)
resh = arr4.reshape(5,5)

print("original array\n", arr4)
print("reshape 5x5\n", resh)
print("second row", resh[1,:])
print("third column", resh[:,2])
print("center matrix\n", resh[1:4, 1:4])
```

6. Given a=np.array([10,20,30,40]) and b= np.array([1,2,3,4]), perform addition, subtraction, multiplication, division, and power(a**2). also compute the dot product of a and b.

```python
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])
dot_pro = np.dot(a, b)
print("addition\n", a + b)
print("subtraction\n", a - b)
print("multiplication\n", a * b)
print("division\n" , a / b)
print("power (a**2)\n", a**2)
print("dot product\n", dot_pro)
```

7. Given arr = np.array([[4,8,2],[9,1,7],[5,6,7]]), find the sum, mean min, max, and standard devation of the whole array, and then the sum along rows(axis = 1) and along columns (axis = 0).

```python
arr5 = np.array([[4, 8, 2],[9, 1, 7],[5, 6, 7]])

print("orignial array\n", arr5)
print("sum", np.sum(arr5))
print("mean", np.mean(arr5))
print("minimum", np.min(arr5))
print("maximum", np.max(arr5))
print("SD", np.std(arr5))
print("sum rows\n", np.sum(arr5, axis=1))
print("sum columns\n", np.sum(arr5, axis=0))
```

8. Create a an array of 20 random integers between 1 and 100. using boolean masking, print (a) all elements greater than 50, (b) all even elements, (c) count of elements divisible by 5.

```python
ran = np.random.randint(1, 101, 20)
div_by5 = np.sum(ran % 5 == 0)
print("array", ran)
print("greater 50", ran[ran > 50])
print("even elements", ran[ran % 2 == 0])
print("divisible by 5:",div_by5 )
```

9. Use np.where to replace all negative values in np.array([5,-3,8,-1,0,-7,4]) with 0. Also the index of the maximum and minimum value using argmax and argmin.

```python
arr6 = np.array([5, -3, 8, -1, 0, -7, 4])
remove_0 = np.where(arr6 < 0, 0, arr6)
print("Updated array", remove_0)
print("maximum index:", np.argmax(remove_0))
print("minimum index:", np.argmin(remove_0))
```

10. Create a 2x3 arrays and join them vertially (vstack) and horizontally (hstack). then sort the array np.array([[9,3,6],[2,8,1]]) along each row and flatten the result into 1D.

```python
a = np.array([[1, 2, 3],[4, 5, 6]])
b = np.array([[7, 8, 9],[10, 11, 12]])

vs_tack = np.vstack((a, b))
hs_tack = np.hstack((a, b))
print("A array\n",a)
print("B array\n", b)

print("vertical stack\n",vs_tack)
print("horizontal stack\n", hs_tack)

arr7 = np.array([[9, 3, 6],[2, 8, 1]])

sort_arr = np.sort(arr7, axis=1)
print("sorted array\n", sort_arr)
print("flattened array:", sort_arr.flatten())
```
