5. Write a program that input a string and ask user to delete a given word from a string.  

```python
sentence = input("Enter a string:")

delete_sentence = input("Enter the word to delete:")

makelst = sentence.split()

if delete_sentence in makelst:
    makelst.remove(delete_sentence)

newsentence = " ".join(makelst)

print("Result:", newsentence)
```

7. Write a program with a function that accepts a string from keyboard and create a new string after converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." the output should be "Stop And Smell The Roses" 

```python
s = input("Enter Sentence")

def cap_words(sen):
    word = sen.split()
    result = ""                      
    for i in word:
        result = result + i.capitalize() + " " 
    return result.strip()           
cap_text = cap_words(s)
print(cap_text)
```

8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be like this:   
[2 11 7 12]  
[5 2 9 15]  
[8 3 10 42] 


Sum of row 1 = 32   
Sum of row 2 = 31   
Sum of row 3 = 63  

```python
m = int(input("Enter Row"))
n = int(input("Enter Column"))

matrix = []

for i in range(m):
    row = []
    for j in range(n):
        value = int(input('Enter element'))
        row.append(value)
    matrix.append(row)
print("Matrix",matrix)

for i in range(m):
    total = 0
    for j in range(n):
        total = total + matrix[i][j]
    print(f"Sum of row {i + 1} = {total}")
```
