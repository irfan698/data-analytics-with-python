person = {

}
Print keys and values using loop

```python
person = {  
"name": "Irfan Arshad",  
"age": 32,   
"city": "Karachi"  
}

for key, value in person.items():
    print(key, ":", value)
```

Q4 — Count Frequency of Numbers
numbers = [1,2,2,3,3,3,4,4,4,4]
Create frequency dictio

```python
num = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

freq = {}
for i in num:
    freq[i] = num.count(i)
print(freq)
```

Q5 — Word Counter
Take sentence input and count word frequency

```python
s = input("Enter sentence")
words = s.split()

freq = {}
for i in words:
    freq[i] = words.count(i)
print('Sentence:',s)
print(freq)
```

Q6 — Student Marks & Average   
marks = {"Math":80, "English":70, "Physics":90}  
Calculate total and average  

```python
marks = {"Math": 80, "English": 70, "Physics": 90}
total = sum(marks.values())
average = total / len(marks)

print("Total:", total)
print("Average:", average)
```

Q9 — Mini Project: Contact Book   
Add 3 contacts, search by name, print phone or Contact not found  

```python
contact = {}
for i in range(3):
    name = input("Enter Name: ")
    phone = input("Enter Phone: ")
    contact[name] = phone
search = input("Search Number: ")
print("Original Data",contact)
if search in contact:
    print("Phone:", contact[search])
else:
    print("Contact not found")
```
