```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://realpython.github.io/fake-jobs/'
request = requests.get(url)
request
```

```python
soup = BeautifulSoup(request.content, 'html.parser') # urlib
soup
```

```python
title = soup.find('h1', class_ = 'title is-1').text.strip()
print(title)
```

```python
job_title = soup.find('h2', class_ = 'title is-5').text.strip
print(job_title)
```

```python
job_subtitle = soup.find('h3', class_ = 'subtitle is-6 company').text.strip()
print(job_subtitle)
```

```python
job_location = soup.find('p', class_ = 'location').text.strip()
print(job_location)
```

```python
job_date = soup.find('p', class_ = 'is-small has-text-grey').text.strip()
print(job_date)
```

```python
job_title = soup.find('h2', class_ = 'title is-5').text.strip()
job_subtitle = soup.find('h3', class_ = 'subtitle is-6 company').text.strip()
job_location = soup.find('p', class_ = 'location').text.strip()
job_date = soup.find('p', class_ = 'is-small has-text-grey').text.strip()
jobs = [job_title, job_subtitle , job_location, job_date]

print(jobs)
```

```python
job_titles = soup.findAll('h2', class_ = 'title is-5')
print(job_titles)
```

```python
for job in job_titles:
    print(job.text.strip())
```

```python
jobs = soup.findAll('div',class_= 'column is-half')
jobs_data = []

for job in jobs:
    job_title = job.find('h2', class_ = 'title is-5').text.strip()
    job_subtitle = job.find('h3', class_ = 'subtitle is-6 company').text.strip()
    job_location = job.find('p', class_ = 'location').text.strip()
    job_date = job.find('p', class_ = 'is-small has-text-grey').text.strip()
    data = [job_title, job_subtitle , job_location, job_date]

    jobs_data.append(data)

print(jobs_data)
```

```python
jobs_data
```

```python
import csv

with open("fake_jobs.csv", mode="w", newline="", encoding="utf-8") as file:
  writer = csv.writer(file)
  writer.writerow(["Job Title", "Company", "Location", "Date"])

  writer.writerows(jobs_data)

print("Data successfully Saved!")




```
